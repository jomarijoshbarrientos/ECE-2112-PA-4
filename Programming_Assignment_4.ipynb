```python
import pandas as pd
import matplotlib.pyplot as plt
```

# A. Visayas Communication Dataframe

```python
df = pd.read_excel('board2.xlsx')
df
```

```python
df['Average'] = df[['Math', 'Electronics', 'GEAS', 'Communication']].mean(axis=1)
VisComm = df.loc[(df['Hometown'] == 'Visayas') &
           (df['Track'] == 'Communication'), ['Name', 'Gender', 'Math', 'Average']]
VisComm
```

```python
print(VisComm)
print('\nNumber of rows:', VisComm.shape[0])
```

# C. Category-Average Visualization

```python
dft = df.groupby('Track')['Average'].mean()
print('Average by Track\n', dft)
```

```python
dfg = df.groupby('Gender')['Average'].mean()
print('Average by Gender\n', dfg)
```

```python
dfh = df.groupby('Hometown')['Average'].mean()
print('Average by Hometown\n', dfh)
```

```python
plt.figure(figsize=(12, 8))

# For Track Values
plt.subplot(1, 3, 1)
plt.bar(dft.index, dft.values, color = 'skyblue')
plt.title('Mean Average by Track')
plt.xlabel('Track')
plt.ylabel('Mean Average')
plt.xticks(rotation=20)

# Gender Values
plt.subplot(1, 3, 2)
plt.bar(dfg.index, dfg.values, color = 'salmon')
plt.title('Mean Average by Gender')
plt.xlabel('Gender')
plt.ylabel('Mean Average')
plt.xticks(rotation=20)

# Hometown Values
plt.subplot(1, 3, 3)
plt.bar(dfh.index, dfh.values, color = 'lightgreen')
plt.title('Mean Average by Hometown')
plt.xlabel('Hometown')
plt.ylabel('Mean Average')
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```
